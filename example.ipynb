# Parquet URL map

Install the included wheel, then replace PARQUET_URL with a spatial Parquet URL accessible from the browser.


```python
import importlib.metadata as metadata
from IPython.display import display
from arcgis_map_compat import Map

print("ipywidgets", metadata.version("ipywidgets"))
print("anywidget", metadata.version("anywidget"))

PARQUET_URL = "https://example.com/data.parquet"
```

```python
m = Map()
m.center = [46.8, 7.16]
m.zoom = 12
m.content.add(PARQUET_URL)
display(m)
```

Use the + and − controls. Then run the next cell to add a second layer to the displayed map. You can replace the URL with another dataset.
