# Minimal notebook map

Install the wheel in this notebook's Python environment. Supply PARQUET_URL through an environment variable or replace the retrieval with your notebook secret store. Do not put literal credentials in cells or print them.

Optional environment variables: FEATURE_SERVICE_URL and ARCGIS_TOKEN; PARQUET_ITEM_ID for an ArcGIS Online Parquet item. Saved output does not retain runtime map definitions; recipients must rerun with their own access.

```python
import os
import importlib.metadata as metadata
from IPython.display import display
from arcgis_map_compat import Map

print('ipywidgets', metadata.version('ipywidgets'))
print('anywidget', metadata.version('anywidget'))
```

```python
m = Map()
m.center = [46.8, 7.16]
m.zoom = 12
m.content.add(os.environ['PARQUET_URL'])
if os.environ.get('FEATURE_SERVICE_URL'):
    m.content.add_feature_service(os.environ['FEATURE_SERVICE_URL'], token=os.environ.get('ARCGIS_TOKEN'))
if os.environ.get('PARQUET_ITEM_ID'):
    m.content.add_parquet_item(os.environ['PARQUET_ITEM_ID'], token=os.environ.get('ARCGIS_TOKEN'))
display(m)
```
